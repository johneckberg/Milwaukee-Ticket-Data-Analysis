# 3. Geocoding using ArcGIS Address Points
To efficiently geocode hundreds of thousands of addresses, we map ticket locations to coordinates from Milwaukee's ArcGIS address dataset.

```python
# load the pre-cleaned address points data (see address_data_processing.ipynb)
milwaukee = pd.read_csv("data/Address_Points_Clean.csv")

# since the geometry was saved to CSV, it is loaded as a string (WKT format). 
# we use GeoPandas to convert it back to actual geometry objects to access .x and .y
milwaukee['geometry'] = gpd.GeoSeries.from_wkt(milwaukee['geometry'])

# build quick lookup dictionary mapping address -> geometry (x/y coordinates)
mke_dict = dict(zip(milwaukee['FullAddr'], milwaukee['geometry']))
missing_set = set()

def geocode_lat(location):
    if location in mke_dict:
        return mke_dict[location].x
    else:
        missing_set.add(location)
        return None
    
def geocode_lng(location):
    if location in mke_dict:
        return mke_dict[location].y
    else:
        missing_set.add(location)
        return None

tickets["Lat"] = tickets["LOCATION"].apply(geocode_lat)
tickets["Lng"] = tickets["LOCATION"].apply(geocode_lng)

# count unrecognized addresses (we will defer dropping them until AFTER the fallback step)
na_num = tickets['Lat'].isna().sum()
print(f"Total Unrecognized Addresses (Exact Match): {na_num}")
print(f"Total Exactly Geocoded Tickets: {len(tickets) - na_num}")
```

### 3b. EDA of Unrecognized Addresses
There is a huge drop in parsed tickets during geocoding. The current model demands an exact string match between the recorded ticket location and our ArcGIS dataset. We can analyze the `missing_set` generated from the lookup function to discover if block-level tracking, intersections, or simple typos caused the mismatch.

```python
import re
import pandas as pd

# analyze the `missing_set` generated during geocoding
missing_df = pd.DataFrame(list(missing_set), columns=['OriginalAddress'])

# categorize the missing addresses based on common patterns
def categorize_missing(address):
    if not isinstance(address, str):
        return "Not a String/NaN"
    address = address.upper()
    if re.search(r'\bBLK\b|\bBLOCK\b|-', address):
        return "Block-Level Address"
    elif re.search(r'/|&|\bAND\b', address):
        return "Intersection"
    elif re.search(r'\bLOT\b', address):
        return "Parking Lot"
    else:
        return "Formatting/Typo/Other"

missing_df['Category'] = missing_df['OriginalAddress'].apply(categorize_missing)

# summary of missing categories
print("Summary of Unrecognized Address Types:")
print(missing_df['Category'].value_counts())
print("\n---")
print("\nSample of Block-Level Addresses:")
print(missing_df[missing_df['Category'] == 'Block-Level Address']['OriginalAddress'].head(5).tolist())
print("\nSample of Intersections:")
print(missing_df[missing_df['Category'] == 'Intersection']['OriginalAddress'].head(5).tolist())
print("\nSample of Formatting/Other Issues:")
print(missing_df[missing_df['Category'] == 'Formatting/Typo/Other']['OriginalAddress'].head(10).tolist())
```

### 3c. Block-Level Geocoding Fallback

Looks like a significant number of them are issues relating to missing house numbers in the city dataset?
To handle this, we can extract the "100-block" (e.g., 1755 -> 1700 block) from both the tickets and our address points. By creating a dictionary that maps each "Street + 100-Block" to its average latitude and longitude, we can trade off exact mapping precision to properly capture these near-misses

```python
# parse Block Keys from the Master Address Dataset
def get_block_key(address):
    """
    Extracts the block and street string from an address:
    '1755 S 37TH ST' -> '17_S 37TH ST'
    '105 N MAIN ST' -> '1_N MAIN ST'
    '500 BLK N MAIN ST' -> '5_N MAIN ST'
    """
    if not isinstance(address, str):
        return None
    # Strip BLK and BLOCK for tickets formatting just in case
    addr = re.sub(r'\bBLK\b|\bBLOCK\b', '', address.upper())
    
    # match the starting numbers and the rest of the street name
    match = re.match(r'^\s*(\d+)\s+(.+)', addr)
    if match:
        house_num = match.group(1)
        street_name = match.group(2).strip()
        
        # floor the house number to the nearest 100 block (1755 // 100 = 17)
        # we store just '17' as the block identifier + the street
        block = int(house_num) // 100
        return f"{block}_{street_name}"
    return None

# duplicate the milwaukee dataset to extract X/Y properties 
block_df = milwaukee.copy()
block_df['BlockKey'] = block_df['FullAddr'].apply(get_block_key)

# drop any that failed to parse a number (e.g. 'MAIN ST')
block_df = block_df.dropna(subset=['BlockKey', 'geometry'])

# The geometry column in the pandas DataFrame does not expose `.x` and `.y` directly.
# We must apply a function to the shapely Point objects to extract them.
block_df['lng'] = block_df['geometry'].apply(lambda pt: pt.x)
block_df['lat'] = block_df['geometry'].apply(lambda pt: pt.y)

# group by our new custom "17_S 37TH ST" keys, and average the lat/long for that entire block
grouped_blocks = block_df.groupby('BlockKey')[['lat', 'lng']].mean()

# convert to a dictionary: '17_S 37TH ST' -> (lat, lng) 
mke_block_dict = grouped_blocks.to_dict('index')
print(f"Created fallback dictionary with {len(mke_block_dict)} street blocks.")

# try the Block keys on the missing Tickets
def fallback_lat(row):
    if pd.notna(row['Lat']): return row['Lat'] # Keep successful exact matches
    key = get_block_key(row['LOCATION'])
    if key in mke_block_dict:
        return mke_block_dict[key]['lat']
    return None

def fallback_lng(row):
    if pd.notna(row['Lng']): return row['Lng'] # Keep successful exact matches
    key = get_block_key(row['LOCATION'])
    if key in mke_block_dict:
        return mke_block_dict[key]['lng']
    return None

# apply the fallback lookup specifically checking empty coordinate columns
start_na = tickets['Lat'].isna().sum()
tickets['Lat'] = tickets.apply(fallback_lat, axis=1)
tickets['Lng'] = tickets.apply(fallback_lng, axis=1)
end_na = tickets['Lat'].isna().sum()

print(f"Reclaimed {start_na - end_na} tickets using the Block-Level average fallback!")

# export the final filtered data
tickets_final = tickets.dropna(subset=['Lat', 'Lng'])
tickets_final.to_csv("Tickets2020LatLngDropNa.csv")
print(f"\nFinal Geocoded Dataset Size: {len(tickets_final)}")
```

```python
# animated interactive scatter map of tickets over time

# only include hover fields that actually exist in the dataframe
# cold probably rremove this?
hover_cols = [c for c in ["ISSUETIME", "VIOCODE", "VIODESCRIPTION", "LOCATION"] if c in plot_df.columns]

fig = px.scatter_mapbox(
    plot_df,
    lat="Lng",   # actual latitude values are stored in Lng
    lon="Lat",   # actual longitude values are stored in Lat
    hover_data=hover_cols,
    animation_frame="ISSUEDATE",
    title="Daily Ticket Dispersal - Milwaukee 2020"
)
fig.update_layout(
    mapbox_style="open-street-map", 
    margin={"r":0,"t":40,"l":0,"b":0},
    mapbox_zoom=10, 
    mapbox_center={"lat": 43.0389, "lon": -87.9065} # center around Milwaukee downtown
)
fig.show()

# download plot as gif
fig.write_html("milwaukee_tickets_2020.html")
```
